```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ast
```

```python
df=pd.read_csv("imdb_data.csv")
```

```python
print(df.shape)
display(df.head())
df.info()
df.describe()
```

```python
print(df.isnull().sum())
```

```python
print("Duplicates:", df.duplicated().sum())
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='budget', y='revenue', data=df, alpha=0.6)
plt.title('Movie Budget vs. Global Revenue')
plt.xlabel('Budget ($)')
plt.ylabel('Revenue ($)')
plt.show()
```

```python
df['genres'] = df['genres'].fillna('[]')
```

```python
df['genres_list'] = df['genres'].apply(ast.literal_eval)
```

```python
df['genre_names'] = df['genres_list'].apply(lambda x: [i['name'] for i in x] if isinstance(x, list) else [])
```

```python
genre_counts = df.explode('genre_names')['genre_names'].value_counts()
```

```python
plt.figure(figsize=(12, 6))
genre_counts.head(10).plot(kind='bar', color='coral')
plt.title('Top 10 Most Frequent Movie Genres')
plt.ylabel('Number of Movies')
plt.xticks(rotation=45)
plt.show()
```

```python
df['profit'] = df['revenue'] - df['budget']
```

```python
df['ROI'] = df['profit'] / (df['budget'] + 0.001)
```

```python
top_profitable = df[['title', 'budget', 'revenue', 'profit']].sort_values(by='profit', ascending=False)
print("--- TOP 5 MOST PROFITABLE MOVIES ---")
print(top_profitable.head(5))
```

```python
df['release_date'] = pd.to_datetime(df['release_date'])
```

```python
df['release_year'] = df['release_date'].dt.year
df['release_month'] = df['release_date'].dt.month
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(x='release_month', y='revenue', data=df, errorbar=None, palette='viridis')
plt.title('Average Revenue by Release Month')
plt.xlabel('Month (1 = Jan, 12 = Dec)')
plt.ylabel('Average Revenue ($)')
plt.show()
```
